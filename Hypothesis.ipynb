#Question 1: What is a null hypothesis (H₀) and why is it important in hypothesis testing?

Ans -

A null hypothesis (H₀) is a statement that assumes no effect, no difference, or no relationship exists in a population. It’s the default claim that researchers test against.

1. It gives you a clear baseline
H₀ assumes “no effect” or “no difference.” Without this starting point, you wouldn’t know what you're comparing your results against.

2. It prevents jumping to conclusions
In Statistics, you don’t immediately accept a claim—you first assume nothing is happening and require strong evidence to prove otherwise. This avoids false claims.

3. It makes results objective
Instead of relying on opinion, decisions are based on data and probability (like p-values). That keeps analysis scientific and unbiased.

4. It helps in decision-making
You use H₀ to decide:

Reject H₀ → something significant is happening
Fail to reject H₀ → not enough evidence

#Question 2: What does the significance level (α) represent in hypothesis testing?

Ans -

The significance level (α) in hypothesis testing is the probability of making a wrong decision by rejecting the null hypothesis (H₀) when it is actually true.

#Question 3: Differentiate between Type I and Type II errors?

Ans -

In hypothesis testing, Type I and Type II errors are two kinds of mistakes you can make when deciding about the null hypothesis (H₀).

🔹 Type I Error (False Positive)
Meaning: Rejecting H₀ when it is actually true
Simple idea: You say “there is an effect” when there is none
Probability: Equal to the significance level (α)
Example:
A test says a person has a disease, but they are actually healthy

🔹 Type II Error (False Negative)
Meaning: Failing to reject H₀ when it is actually false
Simple idea: You say “there is no effect” when there actually is one
Probability: Denoted by β
Example:
A test says a person is healthy, but they actually have the disease

#Question 4: Explain the difference between a one-tailed and two-tailed test. Give an example of each.

Ans -

🔹 One-Tailed Test

A one-tailed test checks for an effect in only one direction (either greater than or less than).

Right-tailed: Tests if a value is greater than a certain point
Left-tailed: Tests if a value is less than a certain point

 You are only interested in one side of the distribution.

Example
H₀: The average score = 50
H₁: The average score > 50

Here, you are only checking if scores increased, not decreased.

A two-tailed test checks for an effect in both directions (greater than or less than).

 You are interested in any difference, whether positive or negative.

Example
H₀: The average score = 50
H₁: The average score ≠ 50

Here, you are checking if scores are either higher or lower than 50.

#Question 6: When should you use a Z-test instead of a t-test?

Ans -

Use a Z-test when:

1.Population standard deviation (σ) is known
This is the main condition.
If σ is given, you use a Z-test directly.

2.Sample size is large (n ≥ 30)
With large samples, the data follows a normal distribution (by Central Limit Theorem), so Z-test works well.

3.Data is approximately normally distributed
Especially important when the sample size is small.
